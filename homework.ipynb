```python
l = [59, 1, 89, 30, 2, 5]
n = len(l)

for i in range(n):
    for j in range(0, n - i - 1):
        if l[j] > l[j + 1]:
            l[j], l[j + 1] = l[j + 1], l[j]

print(l)

print("Sorted List:", l)

print("Minimum:", l[0])
print("Second Smallest:", l[1])
print("Second Largest:", l[-2])
print("Maximum:", l[-1])  
```

```python
def calculator(a, b, op):
    if op == "+":
        return a + b
    elif op == "-":
        return a - b
    elif op == "*":
        return a * b
    elif op == "/":
        return a / b
    else:
        raise ValueError("Invalid Operator")


try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operator = input("Enter operator (+, -, *, /): ")

    result = calculator(num1, num2, operator)
    print("Result:", result)

except ZeroDivisionError:
    print("Error: Division by zero is not allowed")

except ValueError:
    print("Error: Please enter valid input")
```

```python
from collections import Counter

def is_anagram_counter(str1, str2):
     s1 = str1.replace(" ", "").lower()
     s2 = str2.replace(" ", "").lower()
if len(s1) != len(s2):
    return False
return Counter(s1) == Counter(s2)
phrase1 = "A gentleman"
phrase2 = "Elegant man"
if is_anagram_counter(phrase1, phrase2):
    print(f"'{phrase1}' and '{phrase2}' are anagrams.")
else:
    print(f"'{phrase1}' and '{phrase2}' are not anagrams.")
```
